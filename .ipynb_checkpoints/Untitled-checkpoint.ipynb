```python
import shutil, os

files_to_move = [
    "backtest.py", "data_loader.py", "features.py", "main.py",
    "models_sequential.py", "models_tabular.py", "model_arima.py",
    "pipeline.py", "requirements.txt", "README.md"
]

for f in files_to_move:
    src = os.path.join("stock_project", f)
    if os.path.exists(src):
        shutil.copy(src, f)
        print(f"Copied {f}")
    else:
        print(f"Missing: {f}")
```
